# Cursor Verification Notebook

This notebook confirms the basic Cursor workflow by running a small NumPy/Matplotlib example.

```
import numpy as np
import matplotlib.pyplot as plt
```

## Create sine wave data

Use NumPy to build evenly spaced samples and compute the sine values.

```
x = np.linspace(0, 2 * np.pi, 1000)
y = np.sin(x)
```

## Plot the sine wave

Use Matplotlib to render the curve and display the figure.

```
plt.figure(figsize=(8, 4))
plt.plot(x, y, label="sin(x)")
plt.title("Sine Wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True)
plt.legend()
plt.show()
```
